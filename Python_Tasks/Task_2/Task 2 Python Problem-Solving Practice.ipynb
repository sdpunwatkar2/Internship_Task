Q20. Find the Longest Word That Appears Only Once

Question: Find the longest word occurring once, ignoring case; keep the first in a tie.

Observe: Check the given input and expected output.

Understand: Work out what the question asks.

Break: Split the task into small steps.

Logic: Count lowercase words, then compare lengths only for words with frequency one.

Code

Test

```python
sentence = 'Python makes coding useful Python'
words = sentence.split()
normalized = [word.lower() for word in words]
counts = {}
for word in normalized:
    counts[word] = counts.get(word, 0) + 1
longest_unique = None
for word, key in zip(words, normalized):
    if counts[key] == 1:
        if longest_unique is None or len(word) > len(longest_unique):
            longest_unique = word
print(longest_unique)
```

```python
# Additional test case
sentence = 'Apple banana apple mango orange'
words = sentence.split()
normalized = [word.lower() for word in words]
counts = {}
for word in normalized:
    counts[word] = counts.get(word, 0) + 1
longest_unique = None
for word, key in zip(words, normalized):
    if counts[key] == 1:
        if longest_unique is None or len(word) > len(longest_unique):
            longest_unique = word
print(longest_unique)
```
