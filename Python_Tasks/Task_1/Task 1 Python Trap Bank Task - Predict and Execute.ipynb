Q35

prediction
[[1, 99], [3, 4]]

explaination
`copy.copy()` makes a shallow copy: the outer list is new, but the inner lists are shared. Updating an inner list is visible through `a`.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```
